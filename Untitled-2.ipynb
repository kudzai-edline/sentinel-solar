```python
def calculate_bmi(weight, height):
    """
    Calculate BMI and categorize the result.
    
    Parameters:
        weight (float): Weight in kilograms
        height (float): Height in meters
    
    Returns:
        float: The calculated BMI value
    """
    # Validate inputs
    if weight <= 0 or height <= 0:
        raise ValueError("Weight and height must be positive numbers.")
    
    # Calculate BMI
    bmi = weight / (height ** 2)
    
    # Determine category based on World Health Organization standards
    if bmi < 18.5:
        category = "underweight"
    elif bmi < 25:
        category = "normal"
    elif bmi < 30:
        category = "overweight"
    else:
        category = "obese"
    
    # Print the result
    print(f"Your BMI is {bmi:.2f} — you are {category}.")
    
    return bmi


# Example usage
if __name__ == "__main__":
    calculate_bmi(70, 1.75)   # BMI 22.86 → normal
    calculate_bmi(50, 1.75)   # BMI 16.33 → underweight
    calculate_bmi(85, 1.75)   # BMI 27.76 → overweight
    calculate_bmi(100, 1.75)  # BMI 32.65 → obese
```
